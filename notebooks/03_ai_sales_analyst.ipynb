## Product Analysis


Calculate Net Revenue

```python
df["Net_Revenue"] = (
    df["Quantity"]
    * df["Unit_Price"]
    * (1 - df["Discount"] / 100)
)
```

Calculate Cost

```python
df["Total_Cost"] = df["Quantity"] * df["Cost_Price"]
```

Calculate Profit

```python
df["Profit"] = df["Net_Revenue"] - df["Total_Cost"]
```

Creating Product Analysis

```python
product_analysis = (
    df.groupby("Product")
      .agg(
          Orders=("Order_ID", "count"),
          Units_Sold=("Quantity", "sum"),
          Gross_Revenue=("Gross_Revenue", "sum"),
          Net_Revenue=("Net_Revenue", "sum"),
          Profit=("Profit", "sum")
      )
)
```

Calculate Profit Margin

```python
product_analysis["Profit_Margin"] = (
    product_analysis["Profit"]
    / product_analysis["Net_Revenue"]
    * 100
)
product_analysis = product_analysis.sort_values(
    "Net_Revenue",
    ascending=False
)
```

## Multi-Factor Analysis Engine

```python
def select_multiple_analyses(question):
    question = question.lower()

    selected = {}

    if "product" in question:
        selected["Product Analysis"] = ai_analysis_data["top_products"]

    if "category" in question:
        selected["Category Analysis"] = ai_analysis_data["categories"]

    if "region" in question:
        selected["Regional Analysis"] = ai_analysis_data["regions"]

    if "customer" in question:
        selected["Customer Analysis"] = ai_analysis_data["top_customers"]

    if "salesperson" in question:
        selected["Salesperson Analysis"] = ai_analysis_data["salespeople"]

    if "discount" in question:
        selected["Discount Analysis"] = ai_analysis_data["discount_analysis"]

    if (
        "month" in question
        or "monthly" in question
        or "trend" in question
        or "sales increased" in question
        or "sales decreased" in question
    ):
        selected["Monthly Analysis"] = ai_analysis_data["monthly_trends"]

    if (
        "profit" in question
        or "margin" in question
        or "profitable" in question
    ):
        selected["Profitability Analysis"] = ai_analysis_data["profitability"]

    if len(selected) == 0:
        selected["Business Summary"] = ai_analysis_data["business_summary"]

    return selected
```

## Dynamic Business Calculations

```python
def compare_regions(region1, region2):

    r1 = df[df["Region"].str.lower() == region1.lower()]
    r2 = df[df["Region"].str.lower() == region2.lower()]

    revenue1 = r1["Net_Revenue"].sum()
    revenue2 = r2["Net_Revenue"].sum()

    profit1 = r1["Profit"].sum()
    profit2 = r2["Profit"].sum()

    return {
        region1: {
            "Revenue": round(revenue1, 2),
            "Profit": round(profit1, 2)
        },
        region2: {
            "Revenue": round(revenue2, 2),
            "Profit": round(profit2, 2)
        }
    }
```

Dynamic Product Comparison

```python
def compare_products(product1, product2):

    p1 = df[df["Product"].str.lower() == product1.lower()]
    p2 = df[df["Product"].str.lower() == product2.lower()]

    revenue1 = p1["Net_Revenue"].sum()
    revenue2 = p2["Net_Revenue"].sum()

    profit1 = p1["Profit"].sum()
    profit2 = p2["Profit"].sum()

    units1 = p1["Quantity"].sum()
    units2 = p2["Quantity"].sum()

    return {
        product1: {
            "Revenue": round(revenue1, 2),
            "Profit": round(profit1, 2),
            "Units Sold": int(units1)
        },
        product2: {
            "Revenue": round(revenue2, 2),
            "Profit": round(profit2, 2),
            "Units Sold": int(units2)
        }
    }
```

```python
compare_products("Laptop", "Smartphone")
```

AI Recommendation Engine
